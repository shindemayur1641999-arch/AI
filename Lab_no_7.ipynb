```python
# Step 1: Install and import
%pip install -q groq pandas
from google.colab import userdata
from groq import Groq
import time, pandas as pd

# Step 2: LSTM result
lstm_result = {"predicted_category": "Electronics", "confidence": 0.55}
print("LSTM output:", lstm_result)

# Step 3: Send the same query to an LLM
query = "need a smartphone with good camera fast charging under budget five thousand rupee price and best quality and good resolution"
client = Groq(api_key=userdata.get("GROQ_API_KEY"))

start = time.time()
response = client.chat.completions.create(model="openai/gpt-oss-20b",
    messages=[{"role": "user", "content": query}])
latency = time.time() - start

print("\nLLM output:", response.choices[0].message.content)

# Step 4: Calculate token-based cost
cost = (response.usage.prompt_tokens * 0.075 +
        response.usage.completion_tokens * 0.30) / 1000000

```
